```python
import matplotlib.pyplot as plt

country_count = df['Country'].value_counts()

plt.pie(country_count, labels=country_count.index, autopct='%1.0f%%', shadow=True)
plt.title("Chocolate Sales Distribution")
plt.show()
```

Pie chart tells about all the sales of different countries like in India 17% , Australia 20% , New Zealand 15% etc.

```python

top6 = df['Sales Person'].value_counts().head(6).index

plt.figure(figsize=(10,10))
ax = sns.countplot(data=df, x='Sales Person',
                   order=top6)

for bars in ax.containers:
    ax.bar_label(bars)

plt.show()
```

This identifies the top 6 salespersons by frequency and visualizes their number of sales using a Seaborn count plot with value labels displayed on each bar.

```python

top5_amount = df['Amount'].value_counts().nlargest(5).index

df_top5 = df[df['Amount'].isin(top5_amount)]

ax = sns.countplot(data=df_top5, x='Amount', hue='Country')

for bars in ax.containers:
    ax.bar_label(bars)
```

This code selects the top 5 most frequent sales amounts and displays their counts by country using a grouped count plot with value labels on each bar.

```python

Sales_Person = ['Jehu Rudeforth' , 'Van Tuxwell', 'Gigi Bohling ', 'Jan Morforth' , 'Jehu Rudeforth' , 'Van Tuxwell' , 'Oby Sorrel']
Boxes_Shipped = [200, 94, 91, 342, 184, 38, 176]


plt.figure(figsize=(10, 6))
plt.plot(Sales_Person, Boxes_Shipped, marker='o', color='red')  
plt.title("Monthly Sales Trend")
plt.xlabel("Sales Person")
plt.ylabel("Boxes Shipped")
plt.grid(True)
plt.tight_layout()
plt.show()
```

This code creates a line chart showing the number of boxes shipped by each sales person, with markers highlighting individual data points and a grid for better readability.

```python
df_5 = df.head(5)

plt.figure(figsize=(6,4))
sns.boxplot(data=df_5, y='Amount')

plt.title('Chocolate Sales')
plt.ylabel('Sales Amount')
plt.show()
```

This code generates a box plot to visualize the distribution of sales amounts for the first five rows of the chocolate sales dataset.
